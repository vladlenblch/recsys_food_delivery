# Распределение заказов по пользователям

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
user_orders_count = orders.groupby('user_id')['order_number'].max()

print(f"Минимум заказов у юзера: {user_orders_count.min()}")
print(f"Максимум заказов у юзера: {user_orders_count.max()}")

sns.histplot(user_orders_count, bins=97)
plt.title('Распределение общего количества заказов на одного пользователя', fontsize=14)
plt.xlabel('Количество заказов (max order_number)', fontsize=12)
plt.ylabel('Количество пользователей', fontsize=12)
plt.xticks(range(0, 101, 5)) 
plt.show()
```

# Доля заново заказанных товаров

```python
total_reorder_rate = full_data['reordered'].mean()

print(f"Общая доля повторных покупок (reordered=1): {total_reorder_rate:.4f} ({total_reorder_rate * 100:.2f}%)")

reorder_counts = full_data['reordered'].value_counts()
print(f"Новые товары (0): {reorder_counts[0]:,}")
print(f"Повторные (1): {reorder_counts[1]:,}")
```

# Размер каталога и уникальные товары

```python
product_counts = full_data['product_id'].value_counts()

total_unique_products = len(product_counts)
print(f"Всего уникальных product_id в датасете: {total_unique_products:,}")

print("\nСтатистика встречаемости товаров:")
print(f"Куплены ровно 1 раз:  {(product_counts == 1).sum():,} товаров")
print(f"Куплены 5 раз и менее: {(product_counts <= 5).sum():,} товаров")
print(f"Куплены 10 раз и менее: {(product_counts <= 10).sum():,} товаров\n")

def get_catalog_size(min_rep):
    return (product_counts >= min_rep).sum()

for n in range(2, 101):
    remaining = get_catalog_size(n)
    pct = (remaining / total_unique_products) * 100
    print(f"При N >= {n:<3} останется: {remaining:>5,} товаров ({pct:.1f}% каталога)")

thresholds = range(1, 51)
catalog_sizes = [get_catalog_size(n) for n in thresholds]

plt.plot(thresholds, catalog_sizes, marker='o', linewidth=2, markersize=4)
plt.title('Размер каталога в зависимости от минимального порога покупок (N)', fontsize=14)
plt.xlabel('Минимальное количество покупок товара (Порог N)', fontsize=12)
plt.ylabel('Оставшееся количество уникальных товаров', fontsize=12)
plt.xticks(range(1, 52, 5))
plt.show()
```
